```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ', '_')
```

```python
df.columns
```

```python
df=df.drop(columns=['row_number', 'surname'])
```

```python
labels=['Poor', 'Average', 'Good', 'Excellent']
df['credit_score_range']=pd.qcut(df['credit_score'], 4, labels=labels)
```

```python
df[['credit_score', 'credit_score_range']].head(10)
```

```python
age_groups=['Young Adult', 'Adult', 'Middle-Aged', 'Senior']
df['age_group']=pd.qcut(df['age'], 4, labels=age_groups)
```

```python
df[['age', 'age_group']].head(10)
```
